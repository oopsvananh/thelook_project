```python
import os

data_path = "data"

events = pd.read_csv(os.path.join(data_path, "events.csv"))
order_items = pd.read_csv(os.path.join(data_path, "order_items.csv"))
orders = pd.read_csv(os.path.join(data_path, "orders.csv"))
products = pd.read_csv(os.path.join(data_path, "products.csv"))
users = pd.read_csv(os.path.join(data_path, "users.csv"))
```

```python
dict(events['uri'].value_counts())
```

```python
dict(events["uri"].value_counts().head(100))
```

```python
events['uri'].nunique()
```

```python
events['page_type'] = events['uri'].str.extract(r'^/([^/]+)')
events['department'] = events['uri'].str.extract(r'/department/([^/]+)')
events['category'] = events['uri'].str.extract(r'/category/([^/]+)')
events['brand'] = events['uri'].str.extract(r'/brand/([^/]+)')
```

```python
events[10000:20000]
```

Drop một số cột không cần thiết cho analysis

```python
events = events.drop(columns=['ip_address', 'postal_code', 'uri', 'page_type'])
```
